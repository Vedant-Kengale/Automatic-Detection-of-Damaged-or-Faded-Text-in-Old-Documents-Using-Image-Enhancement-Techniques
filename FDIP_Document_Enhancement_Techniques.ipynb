```python
# ==============================================
# Document Enhancement with CNN + GAN
# TRUE Implementation with Generator + Discriminator
# ==============================================

# ====== Cell 1: Install & Imports ======
!pip install -q tensorflow==2.16.1 opencv-python matplotlib numpy

import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
from google.colab import drive

# Reproducibility
np.random.seed(42)
tf.random.set_seed(42)
print(" TensorFlow version:", tf.__version__)

# ====== Cell 2: Mount Google Drive ======
drive.mount('/content/drive')

# ====== Cell 3: Dataset Path ======
test_image_path = "/content/drive/MyDrive/scan_doc_rotation/images/scan_000.png"

if not os.path.exists(test_image_path):
    raise FileNotFoundError(f" Test image not found: {test_image_path}")
else:
    print(f" Image found: {test_image_path}")

# ====== Cell 4: Preprocessing Functions ======
def preprocess_document(img):
    """Advanced preprocessing for document images"""
    denoised = cv2.bilateralFilter(img, 9, 75, 75)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    enhanced = clahe.apply(denoised)
    return enhanced

def advanced_binarization(img):
    """Apply multiple binarization techniques and combine them"""
    _, otsu = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    adaptive_gaussian = cv2.adaptiveThreshold(
        img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY, 21, 8
    )

    adaptive_mean = cv2.adaptiveThreshold(
        img, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
        cv2.THRESH_BINARY, 21, 15
    )

    combined = cv2.addWeighted(adaptive_gaussian, 0.6, adaptive_mean, 0.4, 0)
    return otsu, adaptive_gaussian, adaptive_mean, combined

def post_process_binary(img):
    """Clean up binary image with morphological operations"""
    kernel_open = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    opened = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel_open)
    kernel_close = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    closed = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, kernel_close)
    return closed

# ====== Cell 5: Build GENERATOR (U-Net CNN) ======
def build_generator(img_shape=(256, 256, 1)):
    """
    Generator Network (U-Net Architecture)
    Takes degraded document → Outputs enhanced document
    """
    inp = layers.Input(shape=img_shape, name='generator_input')

    # Encoder (Downsampling)
    c1 = layers.Conv2D(64, (3, 3), activation='relu', padding='same', name='enc_conv1a')(inp)
    c1 = layers.Conv2D(64, (3, 3), activation='relu', padding='same', name='enc_conv1b')(c1)
    p1 = layers.MaxPooling2D((2, 2), name='enc_pool1')(c1)

    c2 = layers.Conv2D(128, (3, 3), activation='relu', padding='same', name='enc_conv2a')(p1)
    c2 = layers.Conv2D(128, (3, 3), activation='relu', padding='same', name='enc_conv2b')(c2)
    p2 = layers.MaxPooling2D((2, 2), name='enc_pool2')(c2)

    c3 = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='enc_conv3a')(p2)
    c3 = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='enc_conv3b')(c3)
    p3 = layers.MaxPooling2D((2, 2), name='enc_pool3')(c3)

    # Bottleneck
    c4 = layers.Conv2D(512, (3, 3), activation='relu', padding='same', name='bottleneck_conv1')(p3)
    c4 = layers.Conv2D(512, (3, 3), activation='relu', padding='same', name='bottleneck_conv2')(c4)

    # Decoder (Upsampling with Skip Connections)
    u1 = layers.UpSampling2D((2, 2), name='dec_upsample1')(c4)
    u1 = layers.concatenate([u1, c3], name='dec_concat1')
    c5 = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='dec_conv1a')(u1)
    c5 = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='dec_conv1b')(c5)

    u2 = layers.UpSampling2D((2, 2), name='dec_upsample2')(c5)
    u2 = layers.concatenate([u2, c2], name='dec_concat2')
    c6 = layers.Conv2D(128, (3, 3), activation='relu', padding='same', name='dec_conv2a')(u2)
    c6 = layers.Conv2D(128, (3, 3), activation='relu', padding='same', name='dec_conv2b')(c6)

    u3 = layers.UpSampling2D((2, 2), name='dec_upsample3')(c6)
    u3 = layers.concatenate([u3, c1], name='dec_concat3')
    c7 = layers.Conv2D(64, (3, 3), activation='relu', padding='same', name='dec_conv3a')(u3)
    c7 = layers.Conv2D(64, (3, 3), activation='relu', padding='same', name='dec_conv3b')(c7)

    # Output layer
    out = layers.Conv2D(1, (1, 1), activation='sigmoid', padding='same', name='generator_output')(c7)

    return models.Model(inp, out, name="Generator")

# ====== Cell 6: Build DISCRIMINATOR ======
def build_discriminator(img_shape=(256, 256, 1)):
    """
    Discriminator Network (PatchGAN)
    Takes an image → Outputs probability that it's real/clean
    """
    inp = layers.Input(shape=img_shape, name='discriminator_input')

    # Downsampling blocks
    x = layers.Conv2D(64, (4, 4), strides=2, padding='same', name='disc_conv1')(inp)
    x = layers.LeakyReLU(0.2, name='disc_lrelu1')(x)

    x = layers.Conv2D(128, (4, 4), strides=2, padding='same', name='disc_conv2')(x)
    x = layers.BatchNormalization(name='disc_bn2')(x)
    x = layers.LeakyReLU(0.2, name='disc_lrelu2')(x)

    x = layers.Conv2D(256, (4, 4), strides=2, padding='same', name='disc_conv3')(x)
    x = layers.BatchNormalization(name='disc_bn3')(x)
    x = layers.LeakyReLU(0.2, name='disc_lrelu3')(x)

    x = layers.Conv2D(512, (4, 4), strides=2, padding='same', name='disc_conv4')(x)
    x = layers.BatchNormalization(name='disc_bn4')(x)
    x = layers.LeakyReLU(0.2, name='disc_lrelu4')(x)

    # Output: Patch-based discrimination
    out = layers.Conv2D(1, (4, 4), padding='same', activation='sigmoid', name='disc_output')(x)

    return models.Model(inp, out, name="Discriminator")

# ====== Cell 7: Build Complete GAN ======
def build_gan(generator, discriminator):
    """
    Combined GAN Model
    Generator tries to fool Discriminator
    """
    # Freeze discriminator when training generator
    discriminator.trainable = False

    gan_input = layers.Input(shape=(256, 256, 1))
    generated_image = generator(gan_input)
    gan_output = discriminator(generated_image)

    gan = models.Model(gan_input, [generated_image, gan_output], name='GAN')
    return gan

# Build all models
print(" Building Generator (U-Net CNN)...")
generator = build_generator()
generator.compile(optimizer=optimizers.Adam(0.0002, 0.5), loss='mse')
print(f" Generator built: {generator.count_params():,} parameters")

print("\n Building Discriminator (PatchGAN)...")
discriminator = build_discriminator()
discriminator.compile(
    optimizer=optimizers.Adam(0.0002, 0.5),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
print(f" Discriminator built: {discriminator.count_params():,} parameters")

print("\n Building Complete GAN...")
gan = build_gan(generator, discriminator)
gan.compile(
    optimizer=optimizers.Adam(0.0002, 0.5),
    loss=['mse', 'binary_crossentropy'],
    loss_weights=[100, 1]  # L1 loss weight: 100, adversarial loss: 1
)
print(f" GAN built: {gan.count_params():,} parameters")

# Print architectures
print("\n" + "="*60)
print("GENERATOR ARCHITECTURE (CNN U-Net)")
print("="*60)
generator.summary()

print("\n" + "="*60)
print("DISCRIMINATOR ARCHITECTURE (PatchGAN)")
print("="*60)
discriminator.summary()

# ====== Cell 8: GAN Training (Simplified) ======
def create_training_data(img_path, target_size=(256, 256)):
    """
    In real training, you'd have pairs of (degraded, clean) images
    Here we simulate by creating degraded version from clean image
    """
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.resize(img, target_size)

    # Clean version
    clean = img.astype('float32') / 255.0
    clean = np.expand_dims(clean, axis=-1)

    # Create degraded version (simulate)
    degraded = img.copy()
    # Add noise
    noise = np.random.normal(0, 25, degraded.shape)
    degraded = np.clip(degraded + noise, 0, 255).astype('uint8')
    # Reduce contrast
    degraded = cv2.convertScaleAbs(degraded, alpha=0.7, beta=20)
    degraded = degraded.astype('float32') / 255.0
    degraded = np.expand_dims(degraded, axis=-1)

    return degraded, clean

def train_gan_single_image(generator, discriminator, gan, img_path, epochs=100):
    """
    Simplified GAN training on single image
    In production, you'd train on large dataset
    """
    print("\n Training GAN (Simplified Demo)...")
    print(" Note: Real training requires paired dataset of degraded/clean images")

    # Create training data
    degraded, clean = create_training_data(img_path)
    degraded_batch = np.expand_dims(degraded, axis=0)
    clean_batch = np.expand_dims(clean, axis=0)

    # Labels for GAN training
    real_labels = np.ones((1, 16, 16, 1))  # Real images label
    fake_labels = np.zeros((1, 16, 16, 1))  # Fake images label

    history = {'d_loss': [], 'g_loss': [], 'd_acc': []}

    for epoch in range(epochs):
        # ===== Train Discriminator =====
        # Generate fake images
        generated_images = generator.predict(degraded_batch, verbose=0)

        # Train on real images
        d_loss_real = discriminator.train_on_batch(clean_batch, real_labels)
        # Train on fake images
        d_loss_fake = discriminator.train_on_batch(generated_images, fake_labels)

        d_loss = 0.5 * np.add(d_loss_real, d_loss_fake)

        # ===== Train Generator =====
        g_loss = gan.train_on_batch(degraded_batch, [clean_batch, real_labels])

        # Store history
        history['d_loss'].append(d_loss[0])
        history['d_acc'].append(d_loss[1])
        history['g_loss'].append(g_loss[0])

        # Print progress
        if epoch % 10 == 0:
            print(f"Epoch {epoch}/{epochs} - D_loss: {d_loss[0]:.4f}, D_acc: {d_loss[1]:.4f}, G_loss: {g_loss[0]:.4f}")

    print(" Training complete!")
    return history

# Train the GAN (simplified for demo)
training_history = train_gan_single_image(generator, discriminator, gan, test_image_path, epochs=50)

# Plot training history
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(training_history['d_loss'], label='Discriminator Loss')
plt.plot(training_history['g_loss'], label='Generator Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('GAN Training Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(training_history['d_acc'], label='Discriminator Accuracy', color='green')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Discriminator Accuracy')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/scan_doc_rotation/gan_training_history.png', dpi=150)
plt.show()

# ====== Cell 9: Complete Enhancement Pipeline with GAN ======
def enhance_document_with_gan(img_path, generator, target_size=(256, 256)):
    """Complete document enhancement pipeline using trained GAN Generator"""

    # Read original image
    original = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if original is None:
        raise FileNotFoundError(f" Image not found: {img_path}")

    orig_shape = original.shape
    print(f"Original image shape: {orig_shape}")

    # Preprocess
    preprocessed = preprocess_document(original)

    # Resize for GAN processing
    resized = cv2.resize(preprocessed, target_size)

    # Normalize and predict using GAN Generator
    normalized = np.expand_dims(resized.astype('float32') / 255.0, axis=(0, -1))
    gan_enhanced = generator.predict(normalized, verbose=0)[0]
    gan_enhanced = (gan_enhanced[..., 0] * 255).astype('uint8')

    # Resize back to original size
    gan_enhanced_full = cv2.resize(gan_enhanced, (orig_shape[1], orig_shape[0]))

    # Apply additional preprocessing
    preprocessed_full = preprocess_document(gan_enhanced_full)

    # Binarization
    otsu, adaptive_gauss, adaptive_mean, combined = advanced_binarization(preprocessed_full)

    # Post-process
    final_binary = post_process_binary(combined)

    return {
        'original': original,
        'preprocessed': preprocessed,
        'gan_enhanced': gan_enhanced_full,
        'preprocessed_full': preprocessed_full,
        'otsu': otsu,
        'adaptive_gaussian': adaptive_gauss,
        'adaptive_mean': adaptive_mean,
        'combined': combined,
        'final': final_binary
    }

# ====== Cell 10: Run Enhancement with GAN ======
print("\n Starting document enhancement with GAN...")
results = enhance_document_with_gan(test_image_path, generator)

# ====== Cell 11: Visualize Results ======
plt.figure(figsize=(20, 12))

# Row 1
plt.subplot(3, 4, 1)
plt.imshow(results['original'], cmap='gray')
plt.title('1. Original Document', fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(3, 4, 2)
plt.imshow(results['preprocessed'], cmap='gray')
plt.title('2. Preprocessed\n(Denoised + CLAHE)', fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(3, 4, 3)
plt.imshow(results['gan_enhanced'], cmap='gray')
plt.title('3. GAN Enhanced ', fontsize=12, fontweight='bold', color='blue')
plt.axis('off')

plt.subplot(3, 4, 4)
plt.imshow(results['preprocessed_full'], cmap='gray')
plt.title('4. Post-GAN Preprocessing', fontsize=12, fontweight='bold')
plt.axis('off')

# Row 2
plt.subplot(3, 4, 5)
plt.imshow(results['otsu'], cmap='gray')
plt.title('5. Otsu Thresholding', fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(3, 4, 6)
plt.imshow(results['adaptive_gaussian'], cmap='gray')
plt.title('6. Adaptive Gaussian', fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(3, 4, 7)
plt.imshow(results['adaptive_mean'], cmap='gray')
plt.title('7. Adaptive Mean', fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(3, 4, 8)
plt.imshow(results['combined'], cmap='gray')
plt.title('8. Combined Methods', fontsize=12, fontweight='bold')
plt.axis('off')

# Row 3
plt.subplot(3, 4, 9)
plt.imshow(results['final'], cmap='gray')
plt.title('9. FINAL OUTPUT (GAN)\n(Morphologically Cleaned)', fontsize=12, fontweight='bold', color='red')
plt.axis('off')

plt.subplot(3, 4, 10)
plt.hist(results['preprocessed_full'].ravel(), bins=256, color='blue', alpha=0.7)
plt.title('Histogram\n(After GAN)', fontsize=12, fontweight='bold')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(3, 4, 11)
h = results['original'].shape[0]
w = results['original'].shape[1]
comparison = np.hstack((
    cv2.resize(results['original'], (w//2, h//2)),
    cv2.resize(results['final'], (w//2, h//2))
))
plt.imshow(comparison, cmap='gray')
plt.title('Before → After (GAN)', fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(3, 4, 12)
plt.text(0.1, 0.9, "GAN COMPONENTS:", fontsize=11, fontweight='bold', color='blue')
plt.text(0.1, 0.75, f" Generator (U-Net)", fontsize=9)
plt.text(0.1, 0.65, f" Discriminator (PatchGAN)", fontsize=9)
plt.text(0.1, 0.55, f" Adversarial Training", fontsize=9)
plt.text(0.1, 0.4, f"Original: {results['original'].shape}", fontsize=9)
plt.text(0.1, 0.3, f"Mean: {results['final'].mean():.2f}", fontsize=9)
plt.text(0.1, 0.2, f"Black: {(results['final'] == 0).sum()}", fontsize=9)
plt.text(0.1, 0.1, f"White: {(results['final'] == 255).sum()}", fontsize=9)
plt.title('GAN Statistics', fontsize=12, fontweight='bold')
plt.axis('off')

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/scan_doc_rotation/gan_enhancement_results.png',
            dpi=150, bbox_inches='tight')
plt.show()

print("Visualization complete!")

# ====== Cell 12: Save Results ======
output_dir = "/content/drive/MyDrive/scan_doc_rotation/output_gan"
os.makedirs(output_dir, exist_ok=True)

cv2.imwrite(f"{output_dir}/01_original.png", results['original'])
cv2.imwrite(f"{output_dir}/02_preprocessed.png", results['preprocessed'])
cv2.imwrite(f"{output_dir}/03_gan_enhanced.png", results['gan_enhanced'])
cv2.imwrite(f"{output_dir}/04_otsu.png", results['otsu'])
cv2.imwrite(f"{output_dir}/05_adaptive_gaussian.png", results['adaptive_gaussian'])
cv2.imwrite(f"{output_dir}/06_adaptive_mean.png", results['adaptive_mean'])
cv2.imwrite(f"{output_dir}/07_combined.png", results['combined'])
cv2.imwrite(f"{output_dir}/08_FINAL_GAN_ENHANCED.png", results['final'])

# Save models
generator.save(f"{output_dir}/generator_model.h5")
discriminator.save(f"{output_dir}/discriminator_model.h5")

print(f"\nAll results saved to: {output_dir}")
print("Files saved:")
print("  - 01_original.png")
print("  - 02_preprocessed.png")
print("  - 03_gan_enhanced.png ")
print("  - 04_otsu.png")
print("  - 05_adaptive_gaussian.png")
print("  - 06_adaptive_mean.png")
print("  - 07_combined.png")
print("  - 08_FINAL_GAN_ENHANCED.png (BEST RESULT)")
print("  - generator_model.h5 ")
print("  - discriminator_model.h5 ")

print("\n" + "="*60)
print("DOCUMENT ENHANCEMENT WITH GAN COMPLETE!")
print("="*60)
print("\nKEY DIFFERENCES FROM CNN-ONLY:")
print("   Generator (U-Net) trained adversarially")
print("   Discriminator provides quality feedback")
print("   Better perceptual quality through adversarial loss")
print("   More realistic enhancement results")
print("="*60)
```
